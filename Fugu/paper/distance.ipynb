```python
from fugu.bricks import SimpleGraphTraversal

# Luckily, they have already created a Brick that we can use to find the distance 
# between two nodes in a graph (Dijkstra's algorithm).
graph = nx.DiGraph()
graph.add_edge(0, 1, weight = 2)
graph.add_edge(0, 2, weight = 5)
graph.add_edge(1, 3, weight = 1)
graph.add_edge(2, 3, weight = 1)

start_node = 0
input = np.zeros(len(graph.nodes))
input[0] = 1

scaffold = Scaffold()
scaffold.add_brick(Vector_Input(input, coding='Undefined', name='input'), 'input')
scaffold.add_brick(SimpleGraphTraversal(graph, name='traversal'), [(0, 0)], output=True)
scaffold.lay_bricks()

backend = snn_Backend()
backend.compile(scaffold, {'record': 'all'})
result = backend.run(50)
print(result)
```

```python
for node, data in scaffold.graph.nodes(data=True):
    print(node, data)
```
